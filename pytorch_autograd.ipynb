```python
x=torch.tensor(6.7)
y=torch.tensor(0.0)
```

```python
w=torch.tensor(1.0,requires_grad=True)
b=torch.tensor(0.0,requires_grad=True)
```

```python
w
```

```python
b
```

```python
z=w*x+b
```

```python
z
```

```python
y_pred=torch.sigmoid(z)
y_pred
```

```python
loss=torch.binary_cross_entropy_with_logits(y_pred,y)
```

```python
loss
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```

# Clearing grad

```python
x=torch.tensor(2.0,requires_grad=True)
x
```

```python
y=x**2
```

```python
y.backward()
y
```

```python
x.grad
```

```python
x.grad.zero_()
```

# Switching off the gradient Tracking

```python
x.requires_grad_(False)
```

```python
x2=torch.tensor(2.0,requires_grad=True)
x2
```

```python
z=x2.detach()
z
```

```python
y=x2**2
y
```

```python
y1=z**2
y1
```

```python
y.backward()
```

```python
x2.grad
```

```python
y1.backward() # beacuase we switch off the gradients using detach function
```
