```python
# create model class

import torch
import torch.nn as nn
```

```python
class Model(nn.Module):

  def __init__(self,num_features):

    super().__init__()
    self.linear=nn.Linear(num_features,1)
    self.sigmoid=nn.Sigmoid()


  def forward(self,features):
    out=self.linear(features)
    out=self.sigmoid(out)

    return out
```

```python
# create datasset
features=torch.rand(10,5)

# create model
model=Model(features.shape[1])

# call model for forward pass
model(features)
```

```python
# show model weights
model.linear.weight
```

```python
# Creating a class for litle complcated model

class Model2(nn.Module):

  def __init__(self,num_features):

    super().__init__()

    self.linear1=nn.Linear(num_features,3)
    self.relu=nn.ReLU()
    self.linear2=nn.Linear(3,1)
    self.sigmiod=nn.Sigmoid()

  def forward(self,features):

    out=self.linear1(features)
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmiod(out)

    return out
```

```python
model2=Model2(features.shape[1])
model2(features)
```

```python
# Creating class for using Sequential container

class ModelContainer(nn.Module):

  def __init__(self,num_features):

    super().__init__()

    self.network=nn.Sequential(
        nn.Linear(num_features,3),
        nn.ReLU(),
        nn.Linear(3,1),
        nn.Sigmoid()
    )

  def forward(self,features):

    out=self.network(features)

    return out
```

```python
model=ModelContainer(features.shape[1])
model(features)
```
